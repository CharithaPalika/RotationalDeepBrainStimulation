```python
import numpy as np
from matplotlib import pyplot as plt
import pickle
```

```python
with open(f'PD\ANALYSIS_DATA.pkl', 'rb') as f:
        DATA = pickle.load(f)
```

```python
SYNC_PD = DATA['SYNC_PD']
ENTROPY_PD = DATA['ENTROPY_PD']
STD_DEV_PD = DATA['STD_DEV_PD']
STD_DEV_Processed_PD = DATA['STD_DEV_Processed_PD']
POWER_BETA_PD = DATA['POWER_BETA_PD']
POWER_BETA_PD_1 = DATA['POWER_BETA_PD_1']
POWER_BETA_PD_2 = DATA['POWER_BETA_PD_2']
# PULSE_INTERVAL_ARRAY = [0, 5, 10, 20, 25]
LAT_STRENGTH_ARRAY = np.linspace(0.0275, 0.05, 10)
KEYS = list(SYNC_PD.keys())
```

```python
fig, ax = plt.subplots(2, 2, figsize = (10, 8))
num = 0
for key in KEYS:
    sync_data = [SYNC_PD[key][i] for i in SYNC_PD[key].keys()]
    # ax[0,0].bar(np.arange(10) + num*0.15, sync_data, label = PULSE_INTERVAL_ARRAY[num], width = 0.15)
    ax[0,0].plot(LAT_STRENGTH_ARRAY, sync_data, marker = 'o')
    ax[0,0].set_title('Synchrony')
    ax[0,0].set_ylim([0, 1])
    # ax[0,0].legend()
    entropy_data = [ENTROPY_PD[key][i] for i in ENTROPY_PD[key].keys()]
    ax[0,1].plot(LAT_STRENGTH_ARRAY, entropy_data, marker = 'o')
    ax[0,1].set_title('Entropy')
    ax[0,1].set_ylim([0, 1])
    # ax[0,1].legend()
    std_processed = [STD_DEV_Processed_PD[key][i].mean() for i in STD_DEV_Processed_PD[key].keys()]
    ax[1,0].plot(LAT_STRENGTH_ARRAY, std_processed, marker = 'o')
    ax[1,0].set_title('Std dev')
    ax[1,0].set_ylim([0, 0.1])
    # ax[1,0].legend()
    power_beta = [POWER_BETA_PD[key][i] for i in POWER_BETA_PD[key].keys()]
    ax[1,1].plot(LAT_STRENGTH_ARRAY, power_beta, marker = 'o')
    ax[1,1].set_title('Power in beta')
    ax[1,1].set_ylim([75, 110])
    ax[1,1].legend()
    num += 1
    # ax[0,0].legend()
plt.tight_layout()
plt.show()
```
