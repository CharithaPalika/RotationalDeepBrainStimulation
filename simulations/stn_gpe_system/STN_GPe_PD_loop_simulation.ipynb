```python
import numpy as np
from matplotlib import pyplot as plt
from analysis import Analysis
from stngpe import STN_GPe_loop
from utils import load_yaml, save_yaml
import scipy
from numpy.fft import fft, ifft
import math
from dbs import *
from tqdm.autonotebook import tqdm
import os
import torch
import pickle
```

```python
def run(lat_strength, yaml_path):
    # yaml_path = r'STD_DBS_PARAMS\params.yaml' 
    arguments = load_yaml(yaml_path)
    arguments['lat_strength_stn'] = float(lat_strength)
    save_yaml(arguments, 'runs.yaml')
    results = STN_GPe_loop('runs.yaml')
    return results
```

```python
YAML_FILES = [r'PD_PARAMS\params.yaml']

DIR_NAMES = ['PD']

LAT_STRENGTH_ARRAY = np.linspace(0.0275, 0.05, 10)
SPIKE_MONITOR_DICT = {}
VOLTAGE_MONITOR_DICT  = {}
LFP_MONITOR_DICT = {}
```

```python
for file_path in tqdm(YAML_FILES):
    for lat_strength in LAT_STRENGTH_ARRAY:
        print(f'Param file: {file_path}, Lateral strength: {lat_strength}')
        results = run(lat_strength=lat_strength, yaml_path=file_path)
        V_stn_time_all = np.array(results['v_stn'])[30000:50000, 15,15] #saving last neuron
        spike_monitor_stn = np.array(results['spike_stn'])[30000:50000]
        lfp_stn = np.array(results['lfp_stn'])[30000:50000]
        SPIKE_MONITOR_DICT[lat_strength] = spike_monitor_stn # last 20000 timesteps
        LFP_MONITOR_DICT[lat_strength] = lfp_stn
        VOLTAGE_MONITOR_DICT[lat_strength] = V_stn_time_all

    # Creating Directory to save results for each param file
    dir_name = DIR_NAMES[YAML_FILES.index(file_path)]

    # Create the directory
    try:
        os.mkdir(os.path.join(os.getcwd(), 'PD', dir_name))
        print(f"Directory '{os.path.join('PD', dir_name)}' created successfully.")
    except FileExistsError:
        print(f"Directory '{os.path.join('PD', dir_name)}' already exists.")

    # Saving results
    with open(os.path.join(os.getcwd(), 'PD', dir_name + r'\spikes.pkl'), 'wb') as file:  
        pickle.dump(SPIKE_MONITOR_DICT, file)


    with open(os.path.join(os.getcwd(), 'PD', dir_name + r'\voltage.pkl'), 'wb') as file:  
        pickle.dump(VOLTAGE_MONITOR_DICT, file)


    with open(os.path.join(os.getcwd(), 'PD',  dir_name + r'\lfp.pkl'), 'wb') as file:  
        pickle.dump(LFP_MONITOR_DICT, file)

    arguments = load_yaml(r'runs.yaml')
    save_yaml(arguments, os.path.join(os.getcwd(), 'PD', dir_name + r'\runs.yaml'))
```
